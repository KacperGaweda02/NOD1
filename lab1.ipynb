```python
# ładowanie biblioteki Pandas
import pandas as pd

# zaimportuj moduł pyplot z biblioteki matplotlib
import matplotlib.pyplot as plt

# wskazać, że wykresy należy rysować bezpośrednio w zeszycie, a nie w osobnej zakładce
%matplotlib inline

# tworzenie ramki danych ze słownika
slownik = {"Kraj": ["Polska", "Niemcy", "Czechy"], "Powszechnosc_Palenia": [0.25, 0.22, 0.28]}
df_slownik = pd.DataFrame(slownik)

# tworzenie ramki danych z listy list
lista_list = [["Polska", 0.25], ["Niemcy", 0.22], ["Czechy", 0.28]]
df_lista = pd.DataFrame(lista_list, columns=["Kraj", "Powszechnosc_Palenia"])

# transponowanie (wymieniamy kolumny a wierszy)
df_transponowane = df_lista.T

# zachowanie ramki danych pobranych z pliku w formacie csv (xlsx)
plik_prev = 'IHME_GBD_2019_SMOKING_AGE_1990_2019_PREV_Y2021M05D27.CSV'
df = pd.read_csv(plik_prev)

# wyświetlić pierwsze 10 wierszy ramki danych
display(df.head(10))

# wyświetlić ostatnie 10 wierszy ramki danych
display(df.tail(10))

# wyświetlić informację o ramce danych
df.info()

# wyświetlić, ile wierszy i kolumn znajduje się w ramce danych
print("Wymiary:", df.shape)

# wyświetlić informację statystyczną o kolumnach liczbowych
display(df.describe())

# wyświetlić informację statystyczną o kolumnach kategoryzowanych
display(df.describe(include=['object']))

# usunąć brakujące wartości w ramce danych
df.dropna(inplace=True)

# przedstawić wybór wierszy i kolumny używając nazw oraz indeksów na różne sposoby
# a) Wybór kolumn po nazwach
wybor_nazwy = df[['location_name', 'sex_name', 'val']]
# b) Wybór wierszy i kolumn przez lokator oparty na etykietach (loc)
wybor_loc = df.loc[10:20, 'location_name':'val']
# c) Wybór wierszy i kolumn przez lokator oparty na indeksach (iloc)
wybor_iloc = df.iloc[10:20, 0:5]

# przedstawić wybór wierszy z ramki danych pod warunkiem odnośnie określonej wartości kolumny
df_mezczyzni = df[df['sex_name'] == 'Male']

# przedstawić wybór wierszy z ramki danych pod warunkiem spełnienia kilku warunków jednocześnie
df_warunki = df[(df['sex_name'] == 'Female') & (df['year_id'].astype(str) == '2019')]

# wybrać wiersze które zawierają w kolumnie kategoryzowanej określone słowo
df_polska = df[df['location_name'].str.contains('Poland', na=False)]

# wybrać wiersze które nie zawierają w kolumnie kategoryzowanej określone słowo
df_nie_polska = df[~df['location_name'].str.contains('Poland', na=False)]

# utwórz kolumnę na podstawie istniejących
df['rozstep_wartosci'] = df['upper'] - df['lower']

# usuń kolumnę
df.drop('rozstep_wartosci', axis=1, inplace=True)

# zmień nazwę kolumny
df.rename(columns={'val': 'wartosc_wskaznika'}, inplace=True)

# zachowaj ramkę danych jako plik csv na komputerze
df.to_csv('moje_wyniki_GBD.csv', index=False)

# wyświetlić średnia (maksymalną, minimalną) wartość z jednej kolumny
print("Średnia:", df['wartosc_wskaznika'].mean())
print("Maksimum:", df['wartosc_wskaznika'].max())
print("Minimum:", df['wartosc_wskaznika'].min())

# wyświetlić liczbę wierszy
print("Liczba wierszy:", len(df))

# wyświetlić wartości unikatowe w kolumnie
print("Unikatowe kraje:", df['location_name'].unique())

# wyświetlić liczby rekordów odpowiadających do wartości
display(df['location_name'].value_counts())

# sortowanie wierszy ramki danych według wartości określonej kolumny (malejąco, rosnąco)
df_malejaco = df.sort_values(by='wartosc_wskaznika', ascending=False)
df_rosnaco = df.sort_values(by='wartosc_wskaznika', ascending=True)

# wyświetlić wierszy dla 10 największych (najmniejszych) wartości określonej kolumny
display(df.nlargest(10, 'wartosc_wskaznika'))
display(df.nsmallest(10, 'wartosc_wskaznika'))

# wyświetlić wierszy dla 10 największych wartości określonej kolumny pod warunkiem określonych wartości innej kolumny
display(df[df['sex_name'] == 'Female'].nlargest(10, 'wartosc_wskaznika'))

# grupowanie wierszy wedlug wartości kolumny kategoryzowanej, potem uśrednienie wartości wszystkich kolumn w grupie MultiIndex
df_zgrupowane = df.groupby(['location_name', 'sex_name']).mean(numeric_only=True)
display(df_zgrupowane)

# grupowanie wierszy wedlug wartości kolumny kategoryzowanej, potem - uśrednienie wartości dla pewnych kolumn, liczba wartości i mediana dla pozostałych kolumn w grupach
df_agg = df.groupby('location_name').agg({
    'wartosc_wskaznika': 'mean',
    'upper': 'median',
    'lower': 'count'
})
display(df_agg)

# wyświetlić nazwy kolumn indeksu złożonego
print(df_zgrupowane.index.names)

# sortować kolumnę indeksu złożonego
display(df_zgrupowane.sort_index())

# stworzyć tabelę przystawną (pivot table) na podstawie ramki danych
tabela_przestawna = df.pivot_table(values='wartosc_wskaznika', index='location_name', columns='sex_name', aggfunc='mean')
display(tabela_przestawna)

# wyświetlić indeksy i kolumny tabeli przystawnej
print("Indeksy tabeli:", tabela_przestawna.index)
print("Kolumny tabeli:", tabela_przestawna.columns)

# utwórz indeks złożony tabeli przystawnej i wyświetl go
tabela_zlozona = df.pivot_table(values='wartosc_wskaznika', index=['location_name', 'age_group_name'], columns='sex_name', aggfunc='mean')
display(tabela_zlozona)

# wyświetlić wykres na podstawie tabeli przystawnej (wybieramy 10 pierwszych, aby uniknąć nakładających się etykiet)
tabela_przestawna.head(10).plot(kind='bar', figsize=(12, 6))
plt.title('Średni wskaźnik powszechności palenia wg płci (próbka)')
plt.ylabel('Powszechność palenia')
plt.xlabel('Kraj')
plt.xticks(rotation=45, ha='right')
plt.show()

# narysować histogram na podstawie wartości kolumny
df['wartosc_wskaznika'].plot(kind='hist', bins=30, figsize=(10, 6), color='skyblue', edgecolor='black')
plt.title('Rozkład wskaźnika powszechności palenia')
plt.xlabel('Wartość wskaźnika')
plt.ylabel('Częstotliwość')
plt.show()

# przedstawić sposoby łączenia ramek danych za pomocą metod merge i concat
# CONCAT
df_czesc1 = df.iloc[:50, :]
df_czesc2 = df.iloc[50:100, :]
df_polaczone_concat = pd.concat([df_czesc1, df_czesc2], axis=0)

# MERGE
plik_init = 'IHME_GBD_2019_SMOKING_AGE_1990_2019_INIT_AGE_Y2021M05D27.CSV'
df_init = pd.read_csv(plik_init)
df_init.rename(columns={'val': 'wiek_inicjacji'}, inplace=True)

# KONIECZNE DLA MERGE: Ujednolicenie typów dla 'year_id' w obu ramkach
df['year_id'] = df['year_id'].astype(str)
df_init['year_id'] = df_init['year_id'].astype(str)

# Łączymy po najbezpieczniejszych wspólnych kluczach
df_polaczone_merge = pd.merge(df, df_init, on=['location_name', 'sex_name', 'year_id'], how='inner')

# pokazać dodawanie nowych kolumn za pomocą operacji matematycznych
srednia_w_zbiorze = df_polaczone_merge['wiek_inicjacji'].mean()
df_polaczone_merge['odchylenie_od_sredniej_inicjacji'] = df_polaczone_merge['wiek_inicjacji'] - srednia_w_zbiorze

# przedstawić na przykładzie dodawanie nowych kolumn z pomocą funkcji lambda
df['czy_mezczyzni'] = df['sex_name'].apply(lambda x: True if x == 'Male' else False)

# przedstawić możliwości pracy z dużymi plikami przy użyciu argumentu chunksize
print("\nPrzykład pracy z argumentem chunksize:")
for fragment_df in pd.read_csv(plik_prev, chunksize=500):
    print("Rozmiar przetworzonego fragmentu (chunk):", fragment_df.shape)
```

```python
import pandas as pd
import numpy as np

# 1. Wczytanie danych z pliku
file_path = "IHME_GBD_2019_SMOKING_AGE_1990_2019_PREV_Y2021M05D27.CSV"
df = pd.read_csv(file_path)

# 2. Średnia arytmetyczna (Mean)
mean_val = np.mean(df["val"])
print(f"Średnia arytmetyczna dla kolumny 'val': {mean_val}")

# 3. Mediana (Median)
median_val = np.median(df["val"])
print(f"Mediana dla kolumny 'val': {median_val}")

# 4. Odchylenie standardowe (Standard Deviation)
std_val = np.std(df["val"])
print(f"Odchylenie standardowe dla kolumny 'val': {std_val}")

# 5. Wariancja (Variance)
variance_val = np.var(df["val"])
print(f"Wariancja dla kolumny 'val': {variance_val}")

# 6. Kowariancja (Covariance)
cov_matrix = np.cov(df["val"], df["upper"])
covariance = cov_matrix[0, 1]
print(f"Kowariancja pomiędzy 'val' a 'upper': {covariance}")

# 7. Korelacja (Correlation)
corr_matrix = np.corrcoef(df["val"], df["upper"])
correlation = corr_matrix[0, 1]
print(f"Korelacja pomiędzy 'val' a 'upper': {correlation}")
```
